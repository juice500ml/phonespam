```python
import praatio.textgrid
def _my_transform(audio_path):
    path = audio_path.replace(".wav", ".TextGrid")
    # Keep empty intervals so the phone rows tile the full audio
    # (leading/trailing silence and internal gaps included).
    grid = praatio.textgrid.openTextgrid(path, includeEmptyIntervals=True)
    tier_name = next(
        x for x in grid.tierNames if x in ("phone", "phones", "Narrow")
    )
    rows = []
    for entry in grid.getTier(tier_name).entries:
        label = (entry.label or "").strip()
        rows.append(
            {
                "min": entry.start,
                "max": entry.end,
                "ipa": label if label else "_",
                "split": "test",
            }
        )
    timestamps = [(row["min"], row["max"]) for row in rows]
    ipas = [row["ipa"] for row in rows]
    if ipas[0] == "_":
        ipas = ipas[1:]
        timestamps = timestamps[1:]
    if ipas[-1] == "_":
        ipas = ipas[:-1]
        timestamps = timestamps[:-1]

    return [t for i, t in enumerate(timestamps) if ipas[i] is not None], [ipa for ipa in ipas if ipa]
```

```python
_IPA_CLOSURE_TO_STOP = {
    "b̚": "b",
    "d̚": "d",
    "ɡ̚": "ɡ",
    "p̚": "p",
    "t̚": "t",
    "k̚": "k",
}
_IPA_AFFRICATE_PAIRS = {
    ("d̚", "d͡ʒ"),
    ("t̚", "t͡ʃ"),
}

IPA_TO_ARPABET = {
    "aʊ": "AW",
    "aɪ": "AY",
    "eɪ": "EY",
    "oʊ": "OW",
    "ɔɪ": "OY",  # until this points not included in powsm vocab
    "t͡ʃ": "CH",
    "d͡ʒ": "JH",
    "ɑ": "AA",
    "æ": "AE",
    "ʌ": "AH",  # unstressed “uh”; schwa is AX
    "ɔ": "AO",
    "ə˞": "AXR",  # r-colored schwa (alternate)
    "ɜ˞": "ER",  # stressed r-colored vowel (alternate)
    "b": "B",
    "d": "D",
    "ð": "DH",
    "ɛ": "EH",
    "f": "F",
    "ɡ": "G",
    "h": "HH",
    "ɪ": "IH",
    "i": "IY",
    "k": "K",
    "l": "L",
    "m": "M",
    "n": "N",
    "ŋ": "NG",
    "p": "P",
    "ɹ": "R",
    "s": "S",
    "ʃ": "SH",
    "t": "T",
    "θ": "TH",
    "ʊ": "UH",
    "u": "UW",
    "v": "V",
    "w": "W",
    "j": "Y",
    "z": "Z",
    "ʒ": "ZH",
    "ə": "AX",
    "ɨ": "IX",
    "l̩": "EL",  # syllabic consonants
    "m̩": "EM",
    "n̩": "EN",
    "ɾ̃": "NX",
    "ɾ": "DX",
    "ʔ": "Q",
    "ʉ": "UX",  # d<u>de
    # -------------------------------
    # TIMIT-SPECIFIC EXTRA SYMBOLS
    # -------------------------------
    "ə̥": "AX-H",  # ax-h (voiceless schwa)
    # Closure symbols (no IPA exact equivalent, best approximation)
    "b̚": "BCL",
    "d̚": "DCL",
    "ɡ̚": "GCL",
    "k̚": "KCL",
    "p̚": "PCL",
    "t̚": "TCL",
    "ŋ̍": "ENG",
    "ʔ̞": "EPI",  #  Glottal or epenthetic event
    "ɦ": "HV",  # voiced /h/
    "ʍ": "WH",
}
ARPABET_TO_IPA = {v.lower(): k for k, v in IPA_TO_ARPABET.items()}

# IPA labels that mark silence/non-speech segments
# after ARPABET→IPA conversion.
IPA_SILENCE_LABELS = {"h#", "pau", "ʔ̞", "sil"}


def process_timit_symbols(
    phone_timestamps,
    phones,
):
    """TIMIT GT pipeline: ARPABET→IPA + closure->stop merge + silence collapse.

    All three steps are idempotent. Labels already in IPA pass through the
    ARPABET map unchanged via the lowercase fallback.

    Args:
        phone_timestamps: list of ``(start, end)`` tuples (seconds).
        phones: parallel list of phone labels (ARPABET or IPA).

    Returns:
        ``(merged_timestamps, merged_phones)`` with phones in IPA, closure-
        stop pairs merged into the stop, and adjacent silence runs
        collapsed into a single span.
    """
    ipa_phones = [ARPABET_TO_IPA.get(p.lower(), p.lower()) for p in phones]
    segs = [(s, e, p) for (s, e), p in zip(phone_timestamps, ipa_phones)]

    merged = []
    i = 0
    while i < len(segs):
        s, e, p = segs[i]
        if p in _IPA_CLOSURE_TO_STOP and i + 1 < len(segs):
            ns, ne, np_ = segs[i + 1]
            if (
                np_ == _IPA_CLOSURE_TO_STOP[p]
                or (p, np_) in _IPA_AFFRICATE_PAIRS
            ):
                merged.append((s, ne, np_))
                i += 2
                continue
        merged.append((s, e, _IPA_CLOSURE_TO_STOP.get(p, p)))
        i += 1

    collapsed = []
    for s, e, p in merged:
        if (
            p in IPA_SILENCE_LABELS
            and collapsed
            and collapsed[-1][2] in IPA_SILENCE_LABELS
        ):
            ps, _, pp = collapsed[-1]
            collapsed[-1] = (ps, e, pp)
            continue
        collapsed.append((s, e, p))
    return (
        [(s, e) for s, e, _ in collapsed],
        ["_" if p in IPA_SILENCE_LABELS else p  for _, _, p in collapsed],
    )
```

```python
import librosa
from phonological_posteriogram import PhoneModel

model = PhoneModel.from_pretrained("models/wavlm-24-timit")
```

```python
import pandas as pd
df = pd.read_csv("data/timit.csv")
```

```python
path = df.iloc[200].audio_path
y = model.load_audio(path)
segs = model.recognize(y, lang="English")

print(len(segs), segs)
df[df.audio_path == path][["ipa", "min", "max"]]
```

```python
import matplotlib.pyplot as plt
import panphon

plt.figure(figsize=(15, 2))
for seg in segs:
    plt.axvline(seg.start, c="black")
    plt.text((seg.start + seg.end)/2, 0.8, seg.label, ha="center", va="center")
plt.axvline(seg.end, c="black")

for _, row in df[df.audio_path == path][["ipa", "min", "max"]].iterrows():
    plt.axvline(row["min"], c="red", ls="--")
    plt.text((row["min"] + row["max"])/2, 0.2, row["ipa"], ha="center", va="center")

_dist = panphon.distance.Distance()
pred = [seg.label for seg in segs]
ref = df[df.audio_path == path]["ipa"].to_list()
cost = _dist.hamming_feature_edit_distance(
    "".join(pred), "".join(ref)
)
print(float(cost) / len(ref))
```
